## Question 1

Create a line plot of monthly sales data.

```python
import matplotlib.pyplot as plt
mon=["jan","feb","mar","apl","may","jun"]
sale=[80,60,80,66,57,87]

plt.plot(mon,sale)
plt.show()
```

## Question 2

Create a bar chart showing marks of 5 students.

```python
stud_name=["s1","s2","s3","s4","s5"]
stud_mark=[60,80,50,85,66]
plt.bar(stud_name,stud_mark)
plt.show()
```

## Question 3

Create a pie chart showing market share of 4 companies.

```python
comp=["tata","jio","mrf","icici"]
share=[30,50,60,40]
ex=[0.1,0,0,0]
plt.pie(share,labels=comp,autopct="%1.1f%%",explode=ex)
plt.show()
```

## Question 4

## Question 5

Create a scatter plot for height vs weight data.

## Question 6

Create a line plot with title and axis labels.

```python
import matplotlib.pyplot as plt
s = ["a","b","c","d"]
m = [10,20,30,40]
plt.plot(s,m, linestyle="--",marker="o",linewidth=1)
plt.title("Line Plot")
plt.xlabel("student")
plt.ylabel("mark")
plt.show()
```

## Question 7

Create a bar chart with custom colors.

```python
stud_name=["s1","s2","s3","s4","s5"]
stud_mark=[60,80,50,85,66]
plt.bar(stud_name,stud_mark,color=['blue','green','black','red','pink'])
plt.show()
```

## Question 8

Create two subplots in one figure.

```python
stud_name=["s1","s2","s3","s4","s5"]
stud_mark=[60,80,50,85,66]
plt.figure(figsize=(12,6))
plt.subplot(1,2,1)
plt.bar(stud_name,stud_mark)

s = ["a","b","c","d"]
m = [10,20,30,40]

plt.subplot(1,2,2)
plt.plot(s,m, linestyle="--",marker="o",linewidth=1)

plt.show()
```

## Question 9

Create a line plot with grid enabled.

```python
mon=["jan","feb","mar","apl","may","jun"]
sale=[50,60,70,80,85,90]

plt.plot(mon,sale)
plt.grid(True)
plt.show()
```

## Question 10

Save a plot as an image file using savefig().

```python
mon=["jan","feb","mar","apl","may","jun"]
sale=[50,60,70,80,85,90]

plt.plot(mon,sale,marker="o")
plt.savefig("chart.png")
plt.show
```
