```python
import matplotlib.pyplot as plt
months = ["Jan","Feb","Mar","Apr"]
sales = [100,150,120,180]
plt.plot(months, sales)
plt.show()
```

```python
import matplotlib.pyplot as plt
students = ["A","B","C","D","E"]
marks = [80,90,75,85,95]
plt.bar(students, marks)
plt.show()
```

```python
import matplotlib.pyplot as plt
shares = [40,30,20,10]
labels = ["A","B","C","D"]
plt.pie(shares, labels=labels, autopct="%1.1f%%")
plt.show()
```

```python
import matplotlib.pyplot as plt
scores = [45,55,60,65,70,75,80,85,90]
plt.hist(scores, bins=5)
plt.show()
```

```python
import matplotlib.pyplot as plt
height = [150,160,170,180]
weight = [50,60,70,80]
plt.scatter(height, weight)
plt.show()
```

```python
import matplotlib.pyplot as plt
x = [1,2,3,4]
y = [10,20,30,40]
plt.plot(x,y)
plt.title("Line Plot")
plt.xlabel("X")
plt.ylabel("Y")
plt.show()
```

```python
import matplotlib.pyplot as plt
students = ["A","B","C"]
marks = [70,85,90]
plt.bar(students, marks, color=["red","green","blue"])
plt.show()
```
