## Question 2

Create a scatter plot showing the relationship between study hours and exam scores.

```python
df=pd.DataFrame({
    "study_time" : [1,2,3,4,5,6,7],
    "exam_scores" : [35,64,87,79,47,64,77]
})
sns.scatterplot(x="study_time",y="exam_scores" , data=df)
plt.show()
```

## Question 3

Create a bar plot showing sales of different products.

```python
df=pd.DataFrame({
    "product" : ['a','b','c','d','e'],
    "sale" : [30,40,30,50,60]
})
sns.barplot(x="product",y="sale",data=df)
plt.show()
```

## Question 4

Create a histogram to visualize the distribution of student marks.

## Question 6

Create a violin plot for student marks grouped by class.

```python
df=pd.DataFrame({
    "class" : ["a","b","c","a","b","c","b","a","c"],
    "mark" : [50,40,66,87,90,67,85,80,98]
})
b=df.groupby('class')["mark"].sum()
print(b)
sns.violinplot(x="class",y="mark" , data=df)
plt.show()
```

## Question 8

Create a heatmap for the correlation matrix of a dataset.

```python
df=pd.DataFrame({
    "s1" : [1,2,3,4,5],
    "s2" : [6,7,8,9,10],
    "s3" : [10,9,11,13,15],
    "s4" : [5,4,3,2,1] 
})
sns.heatmap(df.corr(),annot=True)
plt.show()
```
