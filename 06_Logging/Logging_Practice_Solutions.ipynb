## Question 1

Create a basic logger that prints an INFO message.

```
import logging

logging.basicConfig(level=logging.INFO)
logging.info("This is an INFO message.")
```

## Question 2

Log messages using all five standard logging levels.

```
import logging

logging.basicConfig(level=logging.DEBUG)
logging.debug("Debug message")
logging.info("Info message")
logging.warning("Warning message")
logging.error("Error message")
logging.critical("Critical message")
```

## Question 3

Configure logging to write messages to a file named app.log.

```
import logging

logging.basicConfig(filename="app.log", level=logging.INFO)
logging.info("Application started")
logging.warning("This is a warning")
```

## Question 4

Customize the log format to include time, level, and message.

```
import logging

logging.basicConfig(
    level=logging.INFO,
    format="%(asctime)s - %(levelname)s - %(message)s"
)
logging.info("Formatted log message")
```

## Question 5

Use logging.exception() inside a try-except block.

```
import logging

logging.basicConfig(level=logging.ERROR)

try:
    x = 10 / 0
except Exception:
    logging.exception("An exception occurred")
```

## Question 6

Log user input using the logging module.

```
import logging

logging.basicConfig(level=logging.INFO)

name = input("Enter your name: ")
logging.info(f"User entered: {name}")
```

## Question 7

Create a function that logs when it starts and finishes.

```
import logging

logging.basicConfig(level=logging.INFO)

def process():
    logging.info("Process started")
    print("Processing...")
    logging.info("Process finished")

process()
```

## Question 8

Set the logging level to WARNING and observe which messages appear.

```
import logging

logging.basicConfig(level=logging.WARNING)

logging.debug("Debug")
logging.info("Info")
logging.warning("Warning")
logging.error("Error")
```

## Question 9

Create two loggers with different names and log messages from each.

```
import logging

logger1 = logging.getLogger("ModuleA")
logger2 = logging.getLogger("ModuleB")

logging.basicConfig(level=logging.INFO)

logger1.info("Message from ModuleA")
logger2.info("Message from ModuleB")
```

## Question 10

Use a rotating file handler to limit log file size.

```
import logging
from logging.handlers import RotatingFileHandler

logger = logging.getLogger()
logger.setLevel(logging.INFO)

handler = RotatingFileHandler("rotate.log", maxBytes=500, backupCount=2)
logger.addHandler(handler)

logger.info("Rotating file handler example")
```
