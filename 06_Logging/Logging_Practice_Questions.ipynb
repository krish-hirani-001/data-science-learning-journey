## Question 4

```python
import logging
logging.basicConfig(
    level=logging.DEBUG,
    filename="abc.log",
    format="%(asctime)s - %(levelname)s - %(message)s"  
                    
                    )
logging.debug("Debug Message")
logging.info("Program Started")
logging.warning("Low Disk Space")
logging.error("Unable to Open File")
logging.critical("Database Crashed")
```

## Question 5

Use logging.exception() inside a try-except block.

```python
import logging
logging.basicConfig(
    level=logging.INFO,
    filename="abc.log",
    format="%(asctime)s - %(levelname)s - %(message)s"
)
def division(a,b):
    logging.info("log started")
    try:
        c=a/b
        logging.info(f"addition Completed {a} / {b} = {c}")
        return c
    except Exception as e:
        logging.error(e)
division(20,0)
```

## Question 7

Create a function that logs when it starts and finishes.

```python
import logging
logging.basicConfig(
    level=logging.INFO,
    filename="abc.log",
    format="%(asctime)s - %(levelname)s - %(message)s"
)
def addition(a,b):
    logging.info("log started")
    try:
        c=a/b
        logging.info(f"addition Completed {a} + {b} = {c}")
        return c
    except Exception as e:
        logging.error(e)
    finally:
        logging.info("log finishes")

addition(20,30)
```

## Question 10

Use a rotating file handler to limit log file size.

```python
import logging
from logging.handlers import RotatingFileHandler

logger = logging.getLogger("MyLogger")
logger.setLevel(logging.DEBUG)

handler = RotatingFileHandler(
    "abc.log",
    maxBytes=200,
    backupCount=2
)

formatter = logging.Formatter(
    "%(asctime)s - %(levelname)s - %(message)s"
)

handler.setFormatter(formatter)

logger.addHandler(handler)

for i in range(20):
    logger.info(f"Processing item {i}")
```
