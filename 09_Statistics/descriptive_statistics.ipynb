## 1. Measures of Central Tendency
Types of Measures of Central Tendency <br>
i) Mean <br>
ii) Median<br>
iii) Mode

```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plot
import statistics as st

df=pd.read_csv('employee.csv')
df
```

```python
print(df['age'].mean())
print(df['age'].median())
print(df['age'].mode())
```

```python
# Mean
total=0
for i in df['age']:
    total += i

mean = total / len(df["age"])
print(mean)
```

```python
# Median
a=np.sort(df['age'])
l=len(df['age'])
i=0
if l%2==1:
    b= l//2
    print(a[b])
else:
    b1= l//2 -1
    b2= l//2
    m = (a[b1] + a[b2]) / 2
    print(m)
```

```python
print(st.mean(df['age']))
print(st.median(df['age']))
print(st.mode(df['age']))
```

```python
print(np.mean(df['age']))
print(np.median(df['age']))
```

## 2. Measures of Dispersion

Types of Measures of Dispersion <br>
i) Range <br>
ii) Interquartile Range (IQR) <br>
iii) Mean Absolute Deviation (MAD) <br>
iv) Variance <br>
v) Standard Deviation

```python
df=pd.read_csv('student.csv')
df
```

### i) Range 
FORMULA :- <br>
Range = Maximum Value - Minimum Value

```python
r=df['marks'].max() - df['marks'].min()
print("Range of mark :",r)
```

### ii) Interquartile Range (IQR)

```python
# method 1
q1=df['age'].quantile(0.25)
q3=df['age'].quantile(0.75)
print(f"IQR : {q3-q1}")


# method 2
a=df['age'].describe()['25%']
b=df['age'].describe()['75%']
iqr=b-a
print("IQR : ",iqr)

```

### iii) Mean Absolute Deviation (MAD)


```python
import pandas as pd
import statistics as st
import numpy as np
data=pd.Series([2,4,6,8,10])
a=[]
m=st.mean(data)
for i in data:
    a.append(abs(i-m))
sum=np.sum(a)
l=len(a)
mad=sum/l
print(f"MAD : {mad}")
```

### iv) Variance

```python
print(f"Variance : {st.variance(data)}")
print(np.var(data))
```

### v) Standard Deviation

```python
print(f"std : {st.stdev(data)}")
print(np.std(data))
```

```python
import math
sum_square = 0
data=[2,4,6,8,10]
mean=np.mean(data)
for i in data:
    diff = i - mean
    sum_square += diff ** 2

variance = sum_square / len(data)

standard_deviation = math.sqrt(variance)
print(variance)
print(standard_deviation)
```

## 3. Percentage and Percentile

```python
df=pd.read_csv('student.csv')
df
```

### i) Percentage

```python
df['percentage']=((df['marks'] / 100) * 100).map("{:.2f}%".format)
df.to_csv('student.csv',index=False)
df
```

### ii) Percentile

```python
print("30th percentile :",np.percentile(df['marks'],30))
print("75th percentile :",np.percentile(df['marks'],75))
print("100th percentile :",np.percentile(df['marks'],100))
```

```python
df['percentile']=df['marks'].rank(pct=True) * 100
df.to_csv('student.csv',index=False)
df
```

## 4. Correlation and Covariance

```python
import pandas as pd
data = {
    "hours" : [2,4,6,8],
    "mark" : [40,50,60,70]
}
df=pd.DataFrame(data)
print("\nCovariance :\n",df.cov())

print("\nCorrelation :\n",df.corr())
```

```python
import seaborn as sns
import matplotlib.pyplot as plt
sns.heatmap(df.cov(),annot=True)
plt.show()
```

```python
import seaborn as sns
import matplotlib.pyplot as plt
sns.heatmap(df.corr(),annot=True)
plt.show()
```

## 5. Measures of Shape
i) Positive Skewness (Right Skew) <br>
ii) Negative Skewness (Left Skew) <br>
iii) Symmetric Distribution


```python
df=pd.read_csv('data.csv')
df.head(10)
```

### i) Positive Skewness (Right Skew)

```python
print(df['Annual_Profit_Cr'].skew())
plt.figure(figsize=(6,4))
sns.histplot(x='Annual_Profit_Cr',kde=True,data=df)
plt.title("Positive Skewness (Right Skew)")
plt.show()
```

### ii) Negative Skewness (Left Skew)

```python
print("skew :",df['CEO_Age'].skew())
plt.figure(figsize=(6,4))
sns.histplot(x='CEO_Age',kde=True,data=df)
plt.title("Negative Skewness (Left Skew)")
plt.show()
```

### iii) Symmetric Distribution

```python
print("skew :",df['Employees'].skew())
plt.figure(figsize=(10,4))
sns.histplot(x='Employees',kde=True,data=df)
plt.title("Symmetric Distribution")
plt.show()
```
