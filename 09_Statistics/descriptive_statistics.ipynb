## 1. Measures of Central Tendency
Types of Measures of Central Tendency <br>
i) Mean <br>
ii) Median<br>
iii) Mode

```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plot
import statistics as st

df=pd.read_csv('employee.csv')
df
```

```python
print(df['age'].mean())
print(df['age'].median())
print(df['age'].mode())
```

```python
# Mean
total=0
for i in df['age']:
    total += i

mean = total / len(df["age"])
print(mean)
```

```python
# Median
a=np.sort(df['age'])
l=len(df['age'])
i=0
if l%2==1:
    b= l//2
    print(a[b])
else:
    b1= l//2 -1
    b2= l//2
    m = (a[b1] + a[b2]) / 2
    print(m)
```

```python
print(st.mean(df['age']))
print(st.median(df['age']))
print(st.mode(df['age']))
```

```python
print(np.mean(df['age']))
print(np.median(df['age']))
```
