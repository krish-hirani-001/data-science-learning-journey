## 1. Measures of Central Tendency
Types of Measures of Central Tendency <br>
i) Mean <br>
ii) Median<br>
iii) Mode

```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plot
import statistics as st

df=pd.read_csv('employee.csv')
df
```

```python
print(df['age'].mean())
print(df['age'].median())
print(df['age'].mode())
```

```python
# Mean
total=0
for i in df['age']:
    total += i

mean = total / len(df["age"])
print(mean)
```

```python
# Median
a=np.sort(df['age'])
l=len(df['age'])
i=0
if l%2==1:
    b= l//2
    print(a[b])
else:
    b1= l//2 -1
    b2= l//2
    m = (a[b1] + a[b2]) / 2
    print(m)
```

```python
print(st.mean(df['age']))
print(st.median(df['age']))
print(st.mode(df['age']))
```

```python
print(np.mean(df['age']))
print(np.median(df['age']))
```

## 2. Measures of Dispersion

Types of Measures of Dispersion <br>
i) Range <br>
ii) Interquartile Range (IQR) <br>
iii) Mean Absolute Deviation (MAD) <br>
iv) Variance <br>
v) Standard Deviation

```python
df=pd.read_csv('student.csv')
df
```

### i) Range 
FORMULA :- <br>
Range = Maximum Value - Minimum Value

```python
r=df['marks'].max() - df['marks'].min()
print("Range of mark :",r)
```

### ii) Interquartile Range (IQR)

```python
# method 1
q1=df['age'].quantile(0.25)
q3=df['age'].quantile(0.75)
print(f"IQR : {q3-q1}")


# method 2
a=df['age'].describe()['25%']
b=df['age'].describe()['75%']
iqr=b-a
print("IQR : ",iqr)

```

### iii) Mean Absolute Deviation (MAD)


```python
import pandas as pd
import statistics as st
import numpy as np
data=pd.Series([2,4,6,8,10])
a=[]
m=st.mean(data)
for i in data:
    a.append(abs(i-m))
sum=np.sum(a)
l=len(a)
mad=sum/l
print(f"MAD : {mad}")
```

### iv) Variance

```python
print(f"Variance : {st.variance(data)}")
print(np.var(data))
```

### v) Standard Deviation

```python
print(f"std : {st.stdev(data)}")
print(np.std(data))
```

```python
import math
sum_square = 0
data=[2,4,6,8,10]
mean=np.mean(data)
for i in data:
    diff = i - mean
    sum_square += diff ** 2

variance = sum_square / len(data)

standard_deviation = math.sqrt(variance)
print(variance)
print(standard_deviation)
```
