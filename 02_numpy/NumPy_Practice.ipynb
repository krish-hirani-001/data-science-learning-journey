#### 20. Filtering and Boolean Indexing
##### i) Print all even numbers from an array.<br> ii)Print all odd numbers from an array.<br> iii)Print all elements less than the mean.<br> iv)Replace all values greater than 50 with 100.<br> v) Find elements between 20 and 60.<br> vi) Count how many values are greater than 30.<br> vii) Remove all negative numbers.<br> viii)Replace negative numbers with 0.<br>ix) Print numbers divisible by 5.

```python
a=np.array([1,3,79,2,5,88,7,8,76,6,9,10,55,13,14,12,11,99,15,17,18])
#i)
print("even :",a[a%2==0])
#ii)  
print("\nodd :",a[a%2==1])
#iii)
print("\nelements less than the mean : ",a[a < np.mean(a)])
#iv)
a[a>50]=100
print("\nvalues greater than 50 with 100 :",a)


a=np.array([1,3,79,2,5,88,7,8,76,6,33,9,45,10,55,13,14,12,11,99,15,17,18])
#v)
print("\nelements between 20 and 60 :",a[(a>=20) & (a<=60)])
#vi)
print("\nCount how many values are greater than 30 :",len(a[a>30]))
c=np.array([1,4,6,2,-2,5,-7,-10,11,-14])
#vii)
print("\nRemove all negative numbers :",c[c>0])
#viii)
c[c<0]=0
print("\nReplace negative numbers with 0 :",c)
#ix)
d=np.array([1,10,3,5,20,4,15,7,17,25])
print("\nPrint numbers divisible by 5 :",d[d%5==0])

```
