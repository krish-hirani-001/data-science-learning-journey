# Pandas Data Reading Practice Solutions

## 1. Read students.csv and display all records

```
import pandas as pd
df = pd.read_csv("students.csv")
print(df)
```

## 2. Display only name and age columns

```
df = pd.read_csv("students.csv", usecols=["name","age"])
print(df)
```

## 3. Read first 3 rows from CSV

```
df = pd.read_csv("students.csv", nrows=3)
print(df)
```

## 4. Read students.xlsx

```
df = pd.read_excel("students.xlsx")
print(df)
```

## 5. Read Sheet2 from Excel

```
df = pd.read_excel("students.xlsx", sheet_name="Sheet2")
print(df)
```

## 6. Read students.json

```
df = pd.read_json("students.json")
print(df)
```

## 7. Read first HTML table

```
tables = pd.read_html("students.html")
print(tables[0])
```

## 8. Save DataFrame to Pickle

```
df = pd.read_csv("students.csv")
df.to_pickle("students.pkl")
```

## 9. Read Pickle file

```
df = pd.read_pickle("students.pkl")
print(df)
```

## 10. Compare shape of CSV and JSON data

```
csv_df = pd.read_csv("students.csv")
json_df = pd.read_json("students.json")
print("CSV Shape:", csv_df.shape)
print("JSON Shape:", json_df.shape)
```
