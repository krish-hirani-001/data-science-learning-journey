```python
# Q1
import pandas as pd
import numpy as np
df=pd.DataFrame({'Marks':[90,np.nan,85]})
df.fillna(0)
```

```python
# Q2
df=pd.DataFrame({'Marks':[90,np.nan,85]})
df['Marks']=df['Marks'].fillna(df['Marks'].mean())
df
```

```python
# Q3
df=pd.DataFrame({'Marks':[90,np.nan,85]})
df.ffill()
```

```python
# Q4
df=pd.DataFrame({'Name':['Krish',None],'Marks':[90,80]})
df.dropna()
```

```python
# Q5
df=pd.DataFrame({'Name':['Krish','Raj'],'Marks':[90,None]})
df.dropna(subset=['Marks'])
```

```python
# Q6
df=pd.DataFrame({'Department':['IT','HR','IT'],'Salary':[50000,40000,60000]})
df.groupby('Department')['Salary'].mean()
```

```python
# Q7
df.groupby('Department')['Salary'].sum()
```

```python
# Q8
df.groupby('Department')['Salary'].count()
```

```python
# Q9
df=pd.DataFrame({'Marks':[80,90,70]})
df['Marks']=df['Marks'].apply(lambda x:x+5)
df
```

```python
# Q10
df=pd.DataFrame({'Marks':[95,82,65]})
df['Grade']=df['Marks'].apply(lambda x:'A' if x>=90 else ('B' if x>=75 else 'C'))
df
```

```python
# Q11
import numpy as np
df=pd.DataFrame({'Name':['Krish',None],'Marks':[90,np.nan]})
df['Marks']=df['Marks'].fillna(df['Marks'].mean())
df=df.dropna(subset=['Name'])
df
```

```python
# Q12
df=pd.DataFrame({'Department':['IT','HR','IT'],'Salary':[50000,40000,60000]})
df.groupby('Department')['Salary'].agg(['sum','mean','max','min'])
```

```python
# Q13
df=pd.DataFrame({'Math':[80,90],'Science':[85,95]})
df['Total']=df.apply(lambda row:row['Math']+row['Science'],axis=1)
df
```

```python
# Q14
df=pd.DataFrame({'Name':['Krish'],'Marks':[90]})
df.to_csv('students.csv',index=False)
```

```python
# Q15
import pandas as pd
import numpy as np
df=pd.DataFrame({'Name':['Krish','Raj',None],'Department':['IT','IT','HR'],'Marks':[90,np.nan,75]})
df['Marks']=df['Marks'].fillna(df['Marks'].mean())
df=df.dropna(subset=['Name'])
df['Result']=df['Marks'].apply(lambda x:'Pass' if x>=35 else 'Fail')
report=df.groupby('Department')['Marks'].mean()
df.to_csv('final_report.csv',index=False)
print(report)
```
