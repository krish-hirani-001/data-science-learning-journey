## Q22

```python
dept_df=pd.DataFrame({'Department':['IT','HR','IT','HR'],'Salary':[50000,45000,60000,55000]})
print(dept_df.groupby('Department')['Salary'].sum())
print(dept_df.groupby('Department')['Salary'].mean())
print(dept_df.groupby('Department')['Salary'].max())
```

## Q23

```python
students=pd.DataFrame({'StudentID':[1,2,3],'Name':['A','B','C']})
marks=pd.DataFrame({'StudentID':[1,2,3],'Marks':[80,90,85]})
print(pd.merge(students,marks,on='StudentID'))
```

## Q25

```python
df=pd.DataFrame({'Name':['John','Alice','Bob','Mike'],'Department':['IT','HR','IT','HR'],'Salary':[50000,45000,60000,55000]})
print(df.groupby('Department')['Salary'].mean())
print(df.loc[df.groupby('Department')['Salary'].idxmax()])
print(df.sort_values('Salary',ascending=False).reset_index(drop=True))
```
