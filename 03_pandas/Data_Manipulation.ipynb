## Question 1

Create a DataFrame with missing values and replace them with 0 using fillna().

```python
import numpy as np
import pandas as pd
df=pd.DataFrame({
    "name" : ["krish",np.nan,"raj"],
    "mark" : [50,60,np.nan]
})
print(df)
print("\n",df.fillna(0))
```

## Question 2

Replace missing marks with the column mean.

```python
print(df)
df['mark']=df['mark'].fillna(df['mark'].mean())
print(df)
```

## Question 3

Use forward fill (ffill) to fill missing values.

```python
df=pd.DataFrame({'name' : ['krish',np.nan,'raj'],'mark' : [50,np.nan,60]
})
print(df.ffill())
print(df.bfill())
```

## Question 6

Find average salary department-wise using groupby().

```python
df=pd.DataFrame({
    "name":["krish","yash","meet","ayush","mirav"],
    "age":[20,34,40,30,23],
    "department":["IT","IT","Management","Management","HR"],
    "salary":[30000,45000,35000,23000,43000]
})
print(df.groupby('department')['salary'].mean())
```

## Question 7

Find total salary department-wise using groupby().

```python
print(df.groupby('department')['salary'].sum())
```

## Question 8

Count employees in each department using groupby().

## Question 10

Create a Grade column using apply().

```python
df=pd.DataFrame({"mark":[60,70,50,90,89,95]})
df['grade']=df['mark'].apply(lambda x: 'A' if x>=90 else 'B' if x>=80 else 'c' if x>=60 else 'D')
print(df)
```

## Question 11

Use fillna() and dropna() together on a dataset.

```python
df=pd.DataFrame({
    "name" : ["krish",None,"raj"],
    "mark" : [80,60,np.nan]
})
df['mark']=df['mark'].fillna(df["mark"].mean())
print(df,"\n")
df=df.dropna(subset=["name"])
print(df)
```

## Question 12

Calculate sum, mean, max, and min using groupby().agg().

```python
df=pd.DataFrame({
    "name":["krish","yash","meet","ayush","mirav"],
    "age":[20,34,40,30,23],
    "department":["IT","IT","Management","Management","HR"],
    "salary":[30000,45000,35000,23000,43000]
})
print(df.groupby('department')['salary'].agg(["sum","mean","max","min"]))
```

## Question 15

Create a mini project using fillna(), dropna(), groupby(), apply(), and to_csv().

```python
df=pd.DataFrame({
    "name":["krish","yash",None,"ayush","mirav"],
    "age":[20,34,40,None,23],
    "department":["IT","IT","Management","Management","HR"],
    "salary":[30000,45000,np.nan,23000,43000]
})
df['age']=df["age"].fillna(round(df["age"].mean()))
df['salary']=df["salary"].fillna(df["salary"].mean())
df=df.dropna(subset=['name'])
b=df.groupby('department')["salary"].mean()
df['bonus']=5000
df['total_salary']=df["salary"].apply(lambda x:x+5000)
print(df)
print("\n",b)
df.to_csv('emp1.csv',index=False)
```
