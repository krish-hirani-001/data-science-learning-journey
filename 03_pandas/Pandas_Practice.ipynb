## Q19. Create a DataFrame with NaN values; detect, count, fill, and remove NaNs.

```python
import numpy as np
nan_df=pd.DataFrame({'A':[1,np.nan,3],'B':[4,5,np.nan]})
print(nan_df.isnull())
print("\n",nan_df.isnull().sum())
print("\n",nan_df.fillna(0))
print("\n",nan_df.dropna())
```

## Q22. Group by Department and find total, average, and maximum salary.

```python
dept_df=pd.DataFrame({'Department':['IT','HR','IT','HR'],'Salary':[50000,45000,60000,55000]})
print(dept_df.groupby('Department')['Salary'].sum())
print(dept_df.groupby('Department')['Salary'].mean())
print(dept_df.groupby('Department')['Salary'].max())
```

## Q23. Merge students and marks DataFrames using StudentID.

```python
stud=pd.DataFrame({'iD':[1,2,3],'name':['A','B','C']})
marks=pd.DataFrame({'iD':[1,2,3],'marks':[80,90,85]})
print(pd.merge(stud,marks,on='iD'))
```

## Q25. Group by Department, average salary, highest salary employee, sort descending, and reset index.

```python
df=pd.DataFrame({'Name':['John','Alice','Bob','Mike'],'Department':['IT','HR','IT','HR'],'Salary':[50000,45000,60000,55000]})
print(df.groupby('Department')['Salary'].mean())
print(df.loc[df.groupby('Department')['Salary'].idxmax()])
print(df.sort_values('Salary',ascending=False).reset_index(drop=True))
```
