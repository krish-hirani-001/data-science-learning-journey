### Assignment 1: Handling Division by Zero

Write a function that takes two integers as input and returns their division. Use try, except, and finally blocks to handle division by zero and print an appropriate message.

```python
def abc():
    try:
        a=int(input("enter a number"))
        b=int(input("enter a number"))
        c=a/b
        
    except ZeroDivisionError as e:
        print(e)
        c = None

    finally:
        print("the division is :",c)
abc()
```

### Assignment 5: Exception Handling in Dictionary Access

Write a function that takes a dictionary and a key as input and returns the value associated with the key. Use try, except, and finally blocks to handle KeyError if the key is not found in the dictionary and print an appropriate message.

```python
def get_value(dictionary, key):
    try:
        value = dictionary[key]
        print("Value:", value)

    except KeyError:
        print(f"Error: The key '{key}' was not found in the dictionary.")
        

    finally:
        print("Dictionary operation completed.")
student = {
    "name": "Krish",
    "age": 20,
    "city": "Rajkot"
}

get_value(student, "name")
```

### Assignment 8: Exception Handling in Network Operations

Write a function that attempts to open a URL and read its contents. Use try, except, and finally blocks to handle network-related errors and print an appropriate message.

```python
import requests

def abc(url):
    try:
        response = requests.get(url)
        response.raise_for_status()
        print(response.text)
    
    except requests.RequestException as e:
        print("error :",e)
    
    finally:
        print("Execution complete.")

print(abc('https://jsonplaceholder.typicode.com/posts/1'))
print(abc('https://nonexistent.url'))

```

### Assignment 10: Custom Exception Handling

Define a custom exception named `NegativeNumberError`. Write a function that raises this exception if a negative number is encountered in a list. Use try, except, and finally blocks to handle the custom exception and print an appropriate message.

```python
class NagativeNumberError(Exception):
    pass 

def abc(lst):
    try:
        for num in lst:
            if num < 0:
                raise NagativeNumberError("Negative numbers are not allowed.")
    except NagativeNumberError as e:
        print(e)
    finally:
        print("List processing completed.")
l=[1,2,-3,4,5] 
abc(l)
```

### Assignment 13: Exception Handling in Data Conversion

Write a function that takes a list of strings and converts them to integers. Use try, except, and finally blocks to handle ValueError if a string cannot be converted and print an appropriate message.

```python
def convert_to_integers(str_list):
    integers = []

    try:
        for item in str_list:
            integers.append(int(item))

        print("Converted List:", integers)

    except ValueError:
        print("Error: Cannot convert a string to an integer.")

    finally:
        print("Program execution completed.")


# Example
data = ["10", "20", "abc", "40"]
d1 = ["10", "20", "30", "40"]

convert_to_integers(data)
convert_to_integers(d1)
```

### Assignment 14: Exception Handling in List Comprehensions

Write a function that uses a list comprehension to convert a list of strings to integers. Use try, except, and finally blocks within the list comprehension to handle ValueError and print an appropriate message.

```python
def converted(lst):
    try:
        integers = [int(item) for item in lst]
        print("Converted List:", integers)
    except ValueError as e:
        print(f"Error: {e}")
    finally:
        print("Execution complete.")

converted(data)
converted(d1)
```

### Assignment 15: Exception Handling in File Writing

Write a function that attempts to write a list of strings to a file. Use try, except, and finally blocks to handle IOError and ensure the file is properly closed.

```python
def write_str(str, fn):
    try:
        file = open(fn, 'w')
        for string in str:
            file.write(string + '\n')
    except IOError as e:
        print(f"Error: {e}")
    finally:
        try:
            file.close()
        except NameError:
            pass
write_str(["Hello", "World"], "output.txt")
```
