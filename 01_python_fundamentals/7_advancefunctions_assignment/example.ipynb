## 1. Addition Function

## 2. Even or Odd

#### Write a function that checks whether a number is even or odd.

```python
def even_odd(n):
    if n%2==0:
        print("the number is even")
    else:
        print("the num is odd")
even_odd(14)
```

## 3. Factorial Function

#### Write a function to calculate the factorial of a number.

```python
def fect(n):
    a=1
    for i in range(1,n+1):
        a=a*i
        
    print(a)
fect(5)
```

## 4. Largest Number

#### Write a function that returns the largest of three numbers.

```python
def largst(a,b,c):
    if a>b and a>c:
        print("a is largest")
    elif b>c:
        print("b is largest")
    else:
        print("c is largest")
largst(10,20,30)
```

## 5. Count Vowels

#### Write a function that counts the number of vowels in a string.

```python
def vowel(s):
    c = 0
    for i in s:
        if i in "aeiouAEIOU":
            c = c + 1
    print(c)
vowel("hello world")
```
