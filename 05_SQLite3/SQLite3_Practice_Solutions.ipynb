## Question 2

```
import sqlite3
conn=sqlite3.connect("school.db")
cur=conn.cursor()
cur.execute('''CREATE TABLE IF NOT EXISTS students(
id INTEGER PRIMARY KEY,
name TEXT,
age INTEGER,
marks REAL)''')
conn.commit()
conn.close()
```

## Question 3

```
import sqlite3
conn=sqlite3.connect("school.db")
cur=conn.cursor()
data=[(1,'Alice',20,85),(2,'Bob',21,78),(3,'Charlie',19,88),(4,'David',22,91),(5,'Eva',20,76)]
cur.executemany("INSERT OR REPLACE INTO students VALUES(?,?,?,?)",data)
conn.commit()
conn.close()
```
