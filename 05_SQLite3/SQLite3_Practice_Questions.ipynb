## Question 2

Create a table students with id, name, age, and marks.

```python
import sqlite3
con=sqlite3.connect("school.db")
cursor=con.cursor()
cursor.execute('''
    create table student(
        id INTEGER PRIMARY KEY AUTOINCREMENT,
        name TEXT NOT NULL,
        age INTEGER,
        mark REAL
    )
''')
con.commit()
print("table created...")
```

## Question 3

Insert 5 student records into the students table.

```python
students = [
    ("krish",21,40),
    ("atush",20,66),
    ("meet",18,74),
    ("alice",20,87),
    ("jeet",23,67)
    
]
cursor.executemany("insert into student(name,age,mark) values(?,?,?)",students)
con.commit()
print("record inserted...")
```
